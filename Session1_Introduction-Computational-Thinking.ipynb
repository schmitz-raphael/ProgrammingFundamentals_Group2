## Literals
Literals are notations for representing some fixed values in code.
- `int`. Integers (or simply ints) are one of the numerical types supported by Python. They are numbers written without a fractional component, e.g., `256`, or `-1` (negative integers). 
- floating-point, `float`. They are another one of the numerical types supported by Python. They are numbers that contain (or are able to contain) a fractional component, e.g., `1.27`.
- `String`. Words, texts, numbers formatted as text... 
- Boolean values (binary reptile). Boolean values are the two constant objects True and False used to represent truth values (in numeric contexts 1 is True, while 0 is False.
- `None`. This literal is a `NoneType` object, and it is used to represent the absence of a value.

## Binary System
The binary system is a system of numbers that employs 2 as the base. 
Octal and hexadecimal numeration systems, similarly, employ 8 and 16 as their bases respectively. The hexadecimal system uses the decimal numbers and six extra letters.
Whe do we use them: octal for file permission in Unix, hexadecimal for color codes and memory addresses.
Convenient for conversion from/to binary system.

## Basic operators
- `+` Sum
- `*` Multiplication
- `/` classic division ‒ always returns a float. 
	- ⚠️ Do not divide by zero!
- `//` Floor Division - returns a number resulting from division, but rounded down to the nearest whole number
- `**` Power
- `%` Remainder (modulo). The result of the operator is a remainder left after the integer division.

## Operators and their hierarchy of priorities

|Priority|Operator|Description|Associativity|
|--|--|--|--|
|1|`()`|Parentheses|Left to right|
|2|`x[index], x[index:index]`|Subscription, Slicing|Left to right|
|3|`**`|Exponentiation|Right to left|
|4|`+x`,`-x`,`~x`|Unary plus, Unary minus, Bitwise NOT|Right to left|
|5|`*`,`/`,`//`,`%`|Multiplication, Division, Floor division, Modulus|Left to right|
|6|`+`,`-`|Addition, Subtraction|Left to right|
|7|`<<`,`>>`|Bitwise shift operators|Left to right|
|8|`&`|Bitwise AND|Left to right|
|9|`^`|Bitwise XOR|Left to right|
|10|`\|`|Bitwise OR|Left to right|
|11|`<`,`<=`,`>`,`>=`,`==`,`!=`, `in`, `not in`, `is`, `is not`|Comparisons, Identity, Membership operators|Left to right|
|12|`not`|Boolean NOT|Right to left|
|13|`and`|Boolean AND|Left to right|
|14|`or`|Boolean OR|Left to right|

**NOTE**: A unary operator to the right of the exponentiation operator binds more strongly, for example `4 ** -1` equals `0.25`)

### Operators and their bindings
Most of Python's operators have left-sided binding, which means that the calculation of the expression is conducted from left to right. Some operators act before others. 
- `%` This operator has left-sided binding.
- `**` The exponentiation operator uses right-sided binding.

## Convert Binary, Octal, Decimal, and Hexadecimal

Binary, octal, and hexadecimal are base systems closely related to powers of 2 (binary: $2^1$, octal: $2^3$, hexadecimal: $2^4$).

````python
bin_num = 0b10
oct_num = 0o10
hex_num = 0x10

print(f"The equivalent of 0b10 is {bin_num} and its type is {type(bin_num)}")
print(f"The equivalent of 0o10 is {oct_num} and its type is {type(oct_num)}")
print(f"The equivalent of 0x10 is {hex_num} and its type is {type(hex_num)}\n")

total = 0b10 * 0o10 + 0x10
print(f"These numbers can be subjected to standard arithmetic operations.\nFor example, we can compute 0b10 * 0o10 + 0x10, which output is {total}")
````

In `Python3.6` and later, you can insert underscores `_` within numbers for better readability. Inserting **consecutive underscores raises an error**, but you can include as many non-consecutive underscores as you like.

The underscore can be used as a delimiter for better readability in large numbers. For example, inserting an underscore every four digits can make the number more readable.

````python
print(0b1111_1111_1111)
````

### Convert a number to a binary, octal, and hexadecimal string
- Built-in function `bin()`, `oct()`, `hex()`

````python
n = 8
print(f"The binary form of {n} is {bin(n)} and its type is {type(bin(n))}")
print(f"The octal form of {n} is {oct(n)} and its type is {type(bin(n))}")
print(f"The hexadecimal form of {n} is {hex(n)} and its type is {type(bin(n))}")
````

### Convert a number to a binary, octal, and hexadecimal string
- Built-in function `format()`, string method `str.format()`, f-strings

#### `format()`

````python
print(f"The binary form of {n} is {format(n, '#b')} and its type is {type(format(n, 'b'))}")
print(f"The octal form of {n} is {format(n, '#o')} and its type is {type(format(n, '0'))}")
print(f"The hexadecimal form of {n} is {format(n, '#x')} and its type is {type(format(n, 'x'))}")
````

#### `str.format()`

````python
print('{:08b}'.format(n))
print('{:08o}'.format(n))
print('{:08x}'.format(n))
````

#### `f-strings` 

````python
print("\nf-strings in Python")
print(f"{n:08b}")
print(f"{n:08o}")
print(f"{n:08x}")
````

### Why converting numbers is important? 

````python
i = 5
j = 3

print(format(i, "04b"))
print(format(j, "04b"))

logical_and = bool(i) and bool(j)
print(logical_and)

bitwise_and = i & j
print(format(bitwise_and, "04b"))
````

# 3. Actually Variables: Style Guide for Python Code

## What is a variable?
- A reference, a container, a value that can change.
- A variable is a named location reserved to store values in the memory. Each variable must have a unique name (an identifier).

## Variable names
- The name of the variable **must** begin with a letter.
- The underscore character is a letter.
- upper- and lower-case letters are treated as different (Alice and ALICE are two different variable names in Python, and consequently, two different variables).
- The name of the variable must not be any of Python's **reserved words**. You mustn't use them as names, neither for your variables, functions, or any other named entities you want to create.
```python
['False', 'None', 'True', 'and', 'as', 'assert', 'break', 'class', 'continue', 'def', 'del', 'elif', 'else', 'except', 'finally', 'for', 'from', 'global', 'if', 'import', 'in', 'is', 'lambda', 'nonlocal', 'not', 'or', 'pass', 'raise', 'return', 'try', 'while', 'with', 'yield']
```

https://peps.python.org/pep-0008/
PEP 8 is the official style guide for Python code; gives guidelines on naming conventions, code layout, and other best practices. Ensure that your Python code is readable and maintainable, which helps with collaboration and professional development.


| **Type** | **Naming Convention**| **Description**| **Examples**|
|:----------:|:-------------------|:---------------|:------------|
| Function | Use a lowercase word or words.| Separate words with underscores to improve readability (**snake case**).                     | function, python_function                       |
| Variable* | Use a lowercase single letter, word, or words.  | Separate words with underscores to improve readability (**snake case**).                                | x, var, python_variable. ⚠️ Never use `l`, `O`, or `I` as these can be mistaken for 1 and 0|
| Class    | Start each word with a capital letter.          | Don’t separate words with underscores. This style is called **Pascal case** (e.g., `PascalCase`). | Model, PythonClass                              |
| Method   | Use a lowercase word or words.                  | Separate words with underscores to improve readability (**snake case**).                   | class_method, method                            |
| Constant | Use an uppercase single letter, word, or words. | Separate words with underscores to improve readability.                                | CONSTANT, PYTHON_CONSTANT, PYTHON_LONG_CONSTANT |
| Module   | Use a short, lowercase word or words.           | Separate words with underscores to improve readability.                                | module.py, python_module.py                     |
| Package  | Use a short, lowercase word or words.           | Don’t separate words with underscores.                                                 | package, pythonpackage                          |

### Maximum Line Length and Line Breaking
PEP 8 suggests lines should be limited to **79 characters**.
This allows you to have multiple files open next to one another, while also avoiding line wrapping.
Of course, keeping statements to 79 characters or fewer isn’t always possible. Therefore, PEP 8 also outlines ways to allow statements to run over several lines.
Python will assume line continuation if code is contained within parentheses, brackets, or braces. Example:

````python
def function(arg_one, arg_two,
             arg_three, arg_four):
    return arg_one
````

If you need to break a line around binary operators, like + and *, then you should do so before the operator. This rule stems from mathematics. Mathematicians agree that breaking before binary operators improves readability. 

```python
total = (first_variable
         + second_variable
         - third_variable)
```

## Indentation
- Use **four** consecutive spaces to indicate indentation.
- Prefer spaces over tabs.

While Python code will work with any amount of consistent indentation, four spaces is a widespread convention in the Python community.


## Comments
You should use comments to document code as it’s written. It’s important to document your code so that you, and any collaborators, can understand it. When you or someone else reads a comment, they should be able to easily understand the code that the comment applies to and know how it fits in with the rest of your code.

- Limit the line length of comments and docstrings to 72 characters.
- Use complete sentences, starting with a capital letter.
- Make sure to update comments if you change your code.

You should write your code so that it’s **self-explanatory** whenever possible, and reserve comments for situations where additional explanation is necessary. For example, you might need comments to explain why you wrote your code in a certain way.

**Documentation strings**, or **docstrings**, are strings enclosed in triple double quotation marks (`"""`) or triple single quotation marks (`'''`) that appear on the first line of any function, class, method, or module.
You use docstrings to explain and document a specific block of code. They’re an important part of Python, and you can access the docstring of an object using its `.__doc__` attribute or the `help()` function.

## pip 
`pip` is the package installer for Python developed in 2008. It is the standard tool for installing Python packages and their dependencies in a secure manner. Pip is not just an installer. You can create and manage Python environments, install dependencies, and install packages from third-party repositories using URLs. 

### pip3
Following the release of Python 3, `pip` incorporated the new `pip3` command, which always operates on the Python 3 environment of your computer.


### Installing a list of packages using `pip` requirement files
Create a requirement file, this is usually a text file, that contains all the packages, along with their respective versions, that are used in the project.
```shell
pip install -r requirements.txt
```

## Create a Virtual Environment by using `virtualenv` or `venv`
There are two tools for setting up virtual environments, `virtualenv` and `venv`, that we can use almost interchangeably. `virtualenv` supports older Python versions, in contrast, `venv` is only used with Python 3.3 or higher and is included in the Python standard library, requiring no installation.

### Creating a Python Virtual Environment
```shell
python3 -m venv env_name
```

### Activating a Python Virtual Environment
```bash
source $PATH/env_name/bin/activate
```

### Installing Packages in a Python Virtual Environment
Once inside an isolated virtual environment, check the pre-installed packages on the virtual environment by running the `pip` list command.

```shell
pip list
```

### Reproducing a Python Virtual Environment
The `pip freeze` command allows to create identical environments and to the package list into a `txt` file, run the following command line.
```shell
 pip freeze > requirements.txt
```

### Deleting a Python Virtual Environment
You can simply delete its folder, no uninstall required.
```shell
 rm -rf $PATH/env_name
```

# 5. Version control system
**Git** is a version control system used to manage code. Git allows you to save drafts of your code so that you can look back at previous versions and potentially undo complicated errors. A project managed with Git is called a Git repository<a name="cite_ref-6"></a>[<sup>[6]</sup>](#cite_note-6).

**GitHub** is a popular hosting service for Git repositories. GitHub allows you to store your local Git repositories in the cloud. With GitHub, you can backup your personal files, share your code, and collaborate with others.

## Steps



## Hands-on

- `mkdir hello_git` to make a new directory to practice.
- `cd hello_git` to make the new directory your working directory.
- `git init` to turn the current, empty directory into a fresh Git repository.
- `echo "Hello Git and GitHub" >> README.md` to create a new README file.
- `git add README.md` to add the new file to the Git staging area.
- `git commit -m "First commit"` to make your first commit with the new README file.

### Cloning an existing repository: git clone
Git supports a few different network protocols and corresponding URL formats. In this example, we'll be using the Git SSH protocol. Git SSH URLs follow a template of: `git@HOSTNAME:USERNAME/REPONAME.git`
```shell
git clone <repo url>
```

### Repo-to-repo collaboration: git push


### Git configuration
Git stores configuration options in three separate files, which lets you scope options to individual repositories (local), user (Global), or the entire system (system):

- Local: `/.git/config` => Repository-specific settings.
- Global: `/.gitconfig` => User-specific settings. This is where options set with the --global flag are stored.
- System: `$(prefix)/etc/gitconfig` => System-wide settings.

Define the author name to be used for all commits in the current repository. Typically, you’ll want to use the `--global` flag to set configuration options for the current user.

```bash
git config --global user.name <name>
```

Adding the `--local` option or not passing a config level option at all, will set the user.name for the current local repository.

```bash
git config --local user.email <email>
```
